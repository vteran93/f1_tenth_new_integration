# Experiment Analysis

## 1. Import Required Libraries

```python
import ray
from ray.tune.analysis import ExperimentAnalysis
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from lib.utils import get_experiment_path, get_best_checkpoint
```

## 2. Load and Inspect Experiment Results

```python
experiment_path = get_experiment_path("f1tenth_multiagent_hyper_tuning")
analysis = ExperimentAnalysis(experiment_path)
df = analysis.dataframe()
print(df.shape)
df.head()
```

## 3. Compare Metrics Across Runs

```python
metrics_df = df[["episode_reward_mean", "info/lap_progress", "config/lr", "config/train_batch_size"]].copy()
metrics_df.head()
```

## 4. Visualize Reward Progression

```python
plt.figure(figsize=(12, 6))
sns.lineplot(data=df, x="training_iteration", y="episode_reward_mean", hue="trial_id")
plt.title("Episode Reward Mean vs. Training Iteration")
plt.show()
```

## 5. Visualize Lap Progression

```python
plt.figure(figsize=(12, 6))
sns.lineplot(data=df, x="training_iteration", y="info/lap_progress", hue="trial_id")
plt.title("Lap Progress vs. Training Iteration")
plt.show()
```

## 6. Analyze Hyperparameter Effects

```python
fig, axes = plt.subplots(1, 2, figsize=(15, 6))

sns.scatterplot(data=metrics_df, x="config/lr", y="episode_reward_mean", ax=axes[0])
axes[0].set_title("Episode Reward Mean vs. Learning Rate")

sns.scatterplot(data=metrics_df, x="config/train_batch_size", y="episode_reward_mean", ax=axes[1])
axes[1].set_title("Episode Reward Mean vs. Train Batch Size")

plt.tight_layout()
plt.show()
```

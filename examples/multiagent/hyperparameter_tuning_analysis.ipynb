```python
from pathlib import Path
from ray.tune.analysis import ExperimentAnalysis
import os

# Esto es un path a una carpeta que contiene varios experimentos de Ray Tune
MODELS_PATH = "/home/sergio/repos/f1_tenth_new_integration/examples/multiagent/models_testing/oval_small_PPO_Hyperparameter_Tuning"


experiment = ExperimentAnalysis(experiment_checkpoint_path=os.path.abspath(MODELS_PATH),
                                                          default_metric="env_runners/episode_return_mean",
                                                          default_mode="max")
```

```python
# quiero ver estos parametros:
'''
gamma:            tune.loguniform(0.95, 0.999)     # Log-uniform distribution
  lr:               tune.uniform(1e-5, 1e-3)         # Uniform distribution
  clip_param:       tune.choice([0.1, 0.2, 0.3])     # Discrete choice
  train_batch_size: tune.choice([2000, 4000, 8000])  # Different batch sizes
  num_sgd_iter:     tune.randint(5, 20)              # Random integer
  vf_clip_param:    tune.uniform(0.1, 1.0)  
'''

parameters = ["gamma", "lr", "clip_param", "train_batch_size", "num_epochs", "vf_clip_param"]

for param in parameters:
    try:
      print(f"{param}: {experiment.get_best_config(metric='env_runners/episode_return_mean', mode='max')[param]}")
    except KeyError:
        print(f"Parameter '{param}' not found in the best configuration.")
```

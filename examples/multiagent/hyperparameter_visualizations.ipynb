```python
import os
import numpy as np
import torch
import random
from ray import tune
from ray.tune import CheckpointConfig
from ray.tune.search.optuna import OptunaSearch
from ray.rllib.algorithms.ppo import PPO as PPOTrainer, PPOConfig
from ray.rllib.algorithms.sac import SAC as SACTrainer, SACConfig
from ray.tune.registry import register_env
from ray.rllib.policy.policy import PolicySpec

# Import from lib modules
from lib.utils import load_config, setup_experiment_config, get_reward_class
from lib.multiagent_env import MultiAgentF110

# --- Determinismo ---
os.environ["PYTHONHASHSEED"] = "42"
SEED = 42
np.random.seed(SEED)
torch.manual_seed(SEED)
torch.cuda.manual_seed_all(SEED)
random.seed(SEED)
torch.backends.cudnn.deterministic = True
torch.backends.cudnn.benchmark = False

# --- Load configuration from YAML ---
config_path = "configs/hyperparameter_tuning.yaml"
config_data = load_config(config_path)

# Select experiment to analyze (modify this to select different experiments)
experiment_name = "oval_small_PPO_Hyperparameter_Tuning"  # Change this as needed
experiment = next((exp for exp in config_data['experiments'] if exp['name'] == experiment_name), None)

if not experiment:
    raise ValueError(f"Experiment '{experiment_name}' not found in config")

# Setup experiment configuration
config = setup_experiment_config(experiment, os.path.dirname(config_path))

# Determine algorithm and config
algorithm_name = config['training']['algorithm']
ALGO_MAP = {
    "PPO": (PPOTrainer, PPOConfig, "ppo_params"),
    "SAC": (SACTrainer, SACConfig, "sac_params"),
}

if algorithm_name not in ALGO_MAP:
    raise ValueError(f"Unknown algorithm: {algorithm_name}")

AlgoTrainer, AlgoConfigClass, config_key = ALGO_MAP[algorithm_name]
algo_params = config.get(config_key, {}).copy()

# Extract search space from algorithm parameters (tune functions)
search_space = {}
for key, value in algo_params.items():
    if hasattr(value, '__module__') and 'tune' in str(value.__module__):
        search_space[key] = value

print(f"Selected experiment: {experiment_name}")
print(f"Algorithm: {algorithm_name}")
print(f"Search space parameters: {list(search_space.keys())}")
print(f"Results path: {config['storage_path']}")
print(f"Experiment path: {os.path.join(config['storage_path'], config['name'])}")

# Note: Training execution removed - this notebook is for visualization only
# To run training, use: python run.py train --experiment {experiment_name}
```

```python
from ray.tune import ExperimentAnalysis
import os

# Use the experiment path from the configuration
experiment_path = os.path.join(config["storage_path"], config["name"])

# Check if results exist
if not os.path.exists(experiment_path):
    print(f"❌ Experiment results not found at: {experiment_path}")
    print(f"To run the experiment, use: python run.py train --experiment {experiment_name}")
    print("Available experiments in config:")
    for exp in config_data['experiments']:
        print(f"  - {exp['name']}")
    raise FileNotFoundError(f"No results found for experiment '{experiment_name}'")

print(f"✅ Found experiment results at: {experiment_path}")

# In newer Ray versions, use ExperimentAnalysis to load results
try:
    analysis = ExperimentAnalysis(experiment_path)
    print(f"✅ Successfully loaded experiment analysis")
    print(f"   Number of trials: {len(analysis.trials)}")
    
    # Use the same metric and mode as in your tune.run call
    metric = "env_runners/episode_reward_mean"
    mode = "max"

    # Use the get_best_trial method to get the best trial object
    best_trial = analysis.get_best_trial(metric=metric, mode=mode)
    
    if best_trial:
        print(f"✅ Best trial found with {metric}: {best_trial.last_result.get(metric, 'N/A')}")
        print("Best hyperparameters found were: ", {k: v for k, v in best_trial.config.items() if k in search_space})
        print("Best trial directory: ", best_trial.local_path)
    else:
        print("❌ No best trial found")
        
except Exception as e:
    print(f"❌ Error loading experiment analysis: {e}")
    raise
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Check if analysis is available
if 'analysis' not in locals():
    print("❌ Analysis not loaded. Please run the previous cell first.")
    raise NameError("Analysis not loaded")

# Check if we have trials
if not analysis.trials:
    print("❌ No trials found in the experiment results.")
    raise ValueError("No trials available for visualization")

# --- 1. Get and Display Top 10 Trials ---
all_trials = analysis.trials

# Sort trials by the metric in descending order (for mode="max")
# Handle cases where a trial might not have the metric key
sorted_trials = sorted(
    all_trials,
    key=lambda trial: trial.last_result.get(metric, -float('inf')),
    reverse=True
)

print("--- Top 10 Trials by Reward ---")
top_10_data = []
# Extract only the hyperparameters from the search space for clarity
hyperparam_keys = list(search_space.keys())

for i, trial in enumerate(sorted_trials[:10]):
    data_row = {
        "Rank": i + 1,
        "Reward": trial.last_result.get(metric),
        "Trial ID": trial.trial_id,
    }
    # Add hyperparameter values
    for key in hyperparam_keys:
        data_row[key] = trial.config.get(key)
    top_10_data.append(data_row)

top_10_df = pd.DataFrame(top_10_data)
# Set display options for better readability in the console
pd.set_option('display.max_columns', None)
pd.set_option('display.width', 120)
print(top_10_df)
print("\n" + "="*50 + "\n")


# --- 2. Visualizations with Seaborn (using all trials) ---
# The dataframe is created from analysis.results_df, which contains data from ALL trials.
try:
    df = analysis.results_df
    
    # Clean up column names (e.g., 'config/lr' -> 'lr')
    config_cols = {col: col.replace("config/", "") for col in df.columns if col.startswith("config/")}
    df.rename(columns=config_cols, inplace=True)

    # Define the metric column for easier access
    metric_col = metric
    # Drop rows with NaN rewards if any to avoid plotting errors
    df.dropna(subset=[metric_col], inplace=True)

    if df.empty:
        print("❌ No data available for visualization")
    else:
        print(f"✅ Data loaded: {len(df)} trials available for visualization")
        print("--- Hyperparameter Visualizations (using all trials) ---")
        sns.set_theme(style="whitegrid")

        # Only create visualizations for parameters that exist in search_space
        if "lr" in search_space and "lr" in df.columns:
            # a) Scatter plot for Learning Rate vs. Reward
            plt.figure(figsize=(10, 6))
            sns.scatterplot(data=df, x="lr", y=metric_col, alpha=0.6)
            plt.xscale('log')
            plt.title('Learning Rate vs. Mean Episode Reward (All Trials)')
            plt.xlabel('Learning Rate (log scale)')
            plt.ylabel('Mean Episode Reward')
            plt.show()

        if "train_batch_size" in search_space and "train_batch_size" in df.columns:
            # b) Box plot for a categorical hyperparameter (e.g., train_batch_size) vs. Reward
            plt.figure(figsize=(10, 6))
            # Sort the categories for a cleaner plot
            sorted_batch_sizes = sorted(df['train_batch_size'].unique())
            sns.boxplot(data=df, x='train_batch_size', y=metric_col, order=sorted_batch_sizes)
            plt.title('Train Batch Size vs. Mean Episode Reward (All Trials)')
            plt.xlabel('Train Batch Size')
            plt.ylabel('Mean Episode Reward')
            plt.show()

        # c) Correlation Heatmap
        # Use the actual hyperparameters from search_space
        numeric_hyperparams = [key for key in search_space.keys() 
                              if key in df.columns and df[key].dtype in ['float64', 'int64']]
        # Add the metric column to the list for correlation
        cols_for_corr = numeric_hyperparams + [metric_col]
        # Ensure all selected columns exist in the dataframe
        cols_for_corr = [col for col in cols_for_corr if col in df.columns]

        if len(cols_for_corr) > 1:  # Only create heatmap if we have multiple columns
            plt.figure(figsize=(12, 8))
            corr = df[cols_for_corr].corr()
            sns.heatmap(corr, annot=True, fmt=".2f", cmap="coolwarm", linewidths=.5)
            plt.title('Hyperparameter Correlation Heatmap (All Trials)')
            plt.show()
        else:
            print("ℹ️  Not enough numeric hyperparameters for correlation heatmap")

except Exception as e:
    print(f"❌ Error creating visualizations: {e}")
    raise
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# --- Define the new metric for episode length ---
len_metric = "env_runners/episode_len_mean"
len_mode = "max"  # Assuming longer episodes are better

# --- 1. Get and Display Top 10 Trials by Episode Length ---
# Sort trials by the new metric
sorted_len_trials = sorted(
    analysis.trials,
    key=lambda trial: trial.last_result.get(len_metric, -float('inf')),
    reverse=(len_mode == "max")
)

print(f"--- Top 10 Trials by {len_metric} ---")
top_10_len_data = []
hyperparam_keys = list(search_space.keys())

for i, trial in enumerate(sorted_len_trials[:10]):
    data_row = {
        "Rank": i + 1,
        "Ep. Length": trial.last_result.get(len_metric),
        "Trial ID": trial.trial_id,
    }
    # Add hyperparameter values
    for key in hyperparam_keys:
        data_row[key] = trial.config.get(key)
    top_10_len_data.append(data_row)

top_10_len_df = pd.DataFrame(top_10_len_data)
pd.set_option('display.max_columns', None)
pd.set_option('display.width', 120)
print(top_10_len_df)
print("\n" + "="*50 + "\n")


# --- 2. Visualizations with Seaborn (using all trials) for Episode Length ---
# Use a copy of the results dataframe
df_len = analysis.results_df.copy()

# Clean up column names (e.g., 'config/lr' -> 'lr')
config_cols = {col: col.replace("config/", "") for col in df_len.columns if col.startswith("config/")}
df_len.rename(columns=config_cols, inplace=True)

# Check if the metric column exists before proceeding
if len_metric not in df_len.columns:
    print(f"Metric '{len_metric}' not found in results DataFrame. Skipping visualizations.")
else:
    # Drop rows with NaN values for this metric to avoid plotting errors
    df_len.dropna(subset=[len_metric], inplace=True)

    print(f"--- Hyperparameter Visualizations for {len_metric} (using all trials) ---")
    sns.set_theme(style="whitegrid")

    # a) Scatter plot for Learning Rate vs. Episode Length
    plt.figure(figsize=(10, 6))
    sns.scatterplot(data=df_len, x="lr", y=len_metric, alpha=0.6)
    plt.xscale('log')
    plt.title('Learning Rate vs. Mean Episode Length (All Trials)')
    plt.xlabel('Learning Rate (log scale)')
    plt.ylabel('Mean Episode Length')
    plt.show()

    # b) Box plot for train_batch_size vs. Episode Length
    plt.figure(figsize=(10, 6))
    sorted_batch_sizes = sorted(df_len['train_batch_size'].unique())
    sns.boxplot(data=df_len, x='train_batch_size', y=len_metric, order=sorted_batch_sizes)
    plt.title('Train Batch Size vs. Mean Episode Length (All Trials)')
    plt.xlabel('Train Batch Size')
    plt.ylabel('Mean Episode Length')
    plt.show()

    # c) Correlation Heatmap for Episode Length
    numeric_hyperparams = [
        "lr", "gamma", "lambda_", "clip_param", "train_batch_size",
        "num_sgd_iter", "sgd_minibatch_size", "entropy_coeff"
    ]
    cols_for_corr_len = numeric_hyperparams + [len_metric]
    cols_for_corr_len = [col for col in cols_for_corr_len if col in df_len.columns]

    plt.figure(figsize=(12, 8))
    corr_len = df_len[cols_for_corr_len].corr()
    sns.heatmap(corr_len, annot=True, fmt=".2f", cmap="viridis", linewidths=.5)
    plt.title('Hyperparameter Correlation Heatmap for Mean Episode Length (All Trials)')
    plt.show()
```
